```python
#========================
#TASK 1: DATA IMPORT AND SETUP
#=========================

#Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#------------------------------------------------------------
#LOAD THE DATASET
#------------------------------------------------------------
dataFrame = pd.read_csv("C:\\Users\\reshm\\Downloads\\social_media_engagement_5000.csv")
print("DATASET LOADED SUCCESSFULLY",dataFrame)
```

```python
#Display first 5 rows

print("FIRST 5 ROWS:")
print(dataFrame.head())

#Display last 5 rows

print("LAST 5 ROWS:")
print(dataFrame.tail())

#Dataset shape

print("DATASET SHAPE:")
print(dataFrame.shape)

#Column names

print("COLUMNS:")
print(dataFrame.columns)
```

```python
# ------------------------------------------------------------
# 3. BASIC INFORMATION ABOUT THE DATASET
# ------------------------------------------------------------
print("BASIC INFORMATION ABOUT THE DATASET:")
print("DATA TYPES:")
print(dataFrame.dtypes)

print("DATASET INFORMATION:")
print(dataFrame.info())
```

```python
#-----------------------------------------------
#CONVERT DATE COLUMN TO DATETIME
#-----------------------------------------------

# Convert date columns
dataFrame["posted_at"] = pd.to_datetime(dataFrame["posted_at"],dayfirst=True,errors="coerce")
```

```python
# ------------------------------------------------------------
# 4. CHECKING FOR MISSING VALUES
# ------------------------------------------------------------

print("Missing Values in Each Column:")
print(dataFrame.isnull().sum())
```

```python
dataFrame.describe()
```

```python
# Create a simple NumPy array from likes column
likes_array = np.array(dataFrame['likes'])

# Basic operations
print("First 5 likes:", likes_array[:5])
print("Mean likes:", np.mean(likes_array))
print("Total likes:", np.sum(likes_array))
print("Max likes:", np.max(likes_array))
print("Min likes:", np.min(likes_array))
```

```python
#========================
#TASK 2: DATA CLEANING
#=========================

#---------------------------
#CHECK MISSING VALUES
#---------------------------
print("MISSING VALUES BEFORE CLEANING:")
print(dataFrame.isnull().sum())
```

```python
#------------------------------------------------------------
#REMOVE DUPLICATES
#------------------------------------------------------------

print("NUMBER OF DUPLICATES BEFORE REMOVAL:")
print(dataFrame.duplicated().sum())
df = dataFrame.drop_duplicates()
print("NUMBER OF DUPLICATES AFTER REMOVAL:")
print(dataFrame.duplicated().sum())
```

```python
#-----------------------------------------------
# HANDLE MISSING NUMERICAL VALUES
#-----------------------------------------------

numerical_columns = ["age","likes","comments","shares","watch_time_sec","impression_count",
                     "follower_count","engagement_rate"]

for column in numerical_columns:
    dataFrame[column] = dataFrame[column].fillna(dataFrame[column].median())
```

```python
#------------------------------------------------------------
#HANDLE MISSING CATEGORICAL VALUES
#------------------------------------------------------------

categorical_columns = ["gender","country","post_type","post_category","device_type","sentiment","hashtags"]

for column in categorical_columns:

 if dataFrame[column].isnull().sum() > 0:
    dataFrame[column] = dataFrame[column].fillna(
        dataFrame[column].mode()[0])
```

```python
#------------------------------------------------------------
#HANDLE MISSING DATE VALUES
#------------------------------------------------------------

dataFrame = dataFrame.dropna(
subset=["posted_at"])
```

```python
#------------------------------------------------------------
#STANDARDIZE GENDER VALUES
#------------------------------------------------------------

dataFrame["gender"] = (dataFrame["gender"].astype(str).str.strip().str.lower())
gender_mapping = {"m": "male","male": "male","f": "female","female": "female",
"other": "other"}
dataFrame["gender"] = dataFrame["gender"].replace(gender_mapping)
```

```python
#------------------------------------------------------------
#CLEAN SENTIMENT VALUES
#------------------------------------------------------------

dataFrame["sentiment"] = (dataFrame["sentiment"].astype(str).str.strip().str.lower())
sentiment_mapping = {
"positive": "positive","pos": "positive",
"negative": "negative","neg": "negative",
"neutral": "neutral","neu": "neutral"}
dataFrame["sentiment"] = dataFrame["sentiment"].replace(sentiment_mapping)
```

```python
#-----------------------------------------------
# CORRECT UNREALISTIC NEGATIVE VALUES
#-----------------------------------------------

metrics_columns = ["likes","comments","shares","watch_time_sec","impression_count","follower_count"]

for column in metrics_columns:
    # Replace negative values in the specific column with NaN
    dataFrame.loc[dataFrame[column] < 0, column] = np.nan

    # Fill NaN values with the median of that column
    dataFrame[column] = dataFrame[column].fillna(dataFrame[column].median())
```

```python
#------------------------------------------------------------
#CLEAN AGE VALUES
#------------------------------------------------------------

dataFrame.loc[(dataFrame["age"] < 13) |(dataFrame["age"] > 100),
"age"] = np.nan

dataFrame["age"] = dataFrame["age"].fillna(
dataFrame["age"].median())
```

```python
#------------------------------------------------------------
#CREATE HASHTAG COUNT
#------------------------------------------------------------

dataFrame["hashtag_count"] = (dataFrame["hashtags"].astype(str).str.count("#"))
```

```python
#------------------------------------------------------------
#CHECK MISSING VALUES AFTER CLEANING
#------------------------------------------------------------

print("MISSING VALUES AFTER CLEANING:")

print(dataFrame.isnull().sum())
```

```python
#========================
#TASK 3: DATA EXPLORATION USING PANDAS
#====================================

#------------------------------------------------------------
#DATASET STRUCTURE
#------------------------------------------------------------

print("FIRST 5 ROWS:")
print(dataFrame.head())

print("LAST 5 ROWS:")
print(dataFrame.tail())

print("DATASET SHAPE:")
print(dataFrame.shape)

print("COLUMN NAMES:")
print(dataFrame.columns.tolist())
```

```python
#------------------------------------------------------------
#DATA TYPES
#------------------------------------------------------------

print("DATA TYPES:")
print(dataFrame.dtypes)
```

```python
#------------------------------------------------------------
#SUMMARY STATISTICS
#------------------------------------------------------------

print("SUMMARY STATISTICS:")

print(dataFrame.describe())
```

```python
#------------------------------------------------------------
#CATEGORICAL DISTRIBUTIONS
#------------------------------------------------------------

print("GENDER DISTRIBUTION:")
print(dataFrame["gender"].value_counts())

print("POST TYPE DISTRIBUTION:")
print(dataFrame["post_type"].value_counts())

print("POST CATEGORY DISTRIBUTION:")
print(dataFrame["post_category"].value_counts())

print("COUNTRY DISTRIBUTION:")
print(dataFrame["country"].value_counts())

print("SENTIMENT DISTRIBUTION:")
print(dataFrame["sentiment"].value_counts())
```

```python
#------------------------------------------------------------
#UNIQUE VALUES
#------------------------------------------------------------

print("UNIQUE POST TYPES:")
print(dataFrame["post_type"].unique())

print("NUMBER OF UNIQUE COUNTRIES:")
print(dataFrame["country"].nunique())

print("NUMBER OF UNIQUE POST CATEGORIES:")
print(dataFrame["post_category"].nunique())
```

```python
#------------------------------------------------------------
#CORRELATION MATRIX
#------------------------------------------------------------

numeric_columns_for_corr = ["age","likes","comments","shares","watch_time_sec",
"impression_count","follower_count","engagement_rate","hashtag_count"]

correlation_matrix = dataFrame[numeric_columns_for_corr].corr()

print("CORRELATION MATRIX:")
print(correlation_matrix)
```

```python
#------------------------------------------------------------
#GROUPBY: AVERAGE LIKES BY POST TYPE
#------------------------------------------------------------

avg_likes_post_type = (dataFrame.groupby("post_type")["likes"].mean()
.sort_values(ascending=False))

print("AVERAGE LIKES BY POST TYPE:")
print(avg_likes_post_type)
```

```python
#------------------------------------------------------------
#GROUPBY: AVERAGE IMPRESSIONS BY COUNTRY
#------------------------------------------------------------

avg_impressions_country = (dataFrame.groupby("country")["impression_count"].mean()
.sort_values(ascending=False))

print("AVERAGE IMPRESSIONS BY COUNTRY:")
print(avg_impressions_country)
```

```python
#========================
#TASK 4: DATA WRANGLING
#=========================

#------------------------------------------------------------
#CREATE ENGAGEMENT SCORE
#------------------------------------------------------------

dataFrame["engagement_score"] = (dataFrame["likes"] +dataFrame["comments"] +dataFrame["shares"])
```

```python
#------------------------------------------------------------
#LOG TRANSFORMATION
#-----------------------------------------------------------

dataFrame["log_likes"] = np.log1p(dataFrame["likes"])

dataFrame["log_impressions"] = np.log1p(dataFrame["impression_count"])
```

```python
#------------------------------------------------------------
#GROUPBY BY POST TYPE
#------------------------------------------------------------

post_type_summary = (dataFrame.groupby("post_type").agg(
    Average_Likes=("likes", "mean"),
    Average_Comments=("comments", "mean"),
    Average_Shares=("shares", "mean"),
    Average_Engagement=("engagement_score", "mean"),
    Post_Count=("post_id", "count"))
.sort_values("Average_Engagement",ascending=False))

print("POST TYPE SUMMARY:")
print(post_type_summary)
```

```python
#------------------------------------------------------------
#GROUPBY BY COUNTRY
#------------------------------------------------------------

country_summary = (dataFrame.groupby("country").agg(
    Average_Engagement_Rate=("engagement_rate","mean"),
    Average_Engagement=("engagement_score","mean"))
.sort_values("Average_Engagement_Rate",ascending=False))

print("COUNTRY SUMMARY:")
print(country_summary)
```

```python
#------------------------------------------------------------
#GROUPBY BY SENTIMENT
#------------------------------------------------------------

sentiment_summary = (dataFrame.groupby("sentiment").agg(
    Average_Likes=("likes","mean"),
    Average_Engagement=("engagement_score","mean"),
    Average_Engagement_Rate=("engagement_rate","mean"))
.sort_values("Average_Engagement",ascending=False))

print("SENTIMENT SUMMARY:")
print(sentiment_summary)
```

```python
#========================
#TASK 5: STATISTICAL ANALYSIS
#========================

statistics_columns = ["likes","comments","shares","watch_time_sec",
"engagement_rate","follower_count"]

for column in statistics_columns:
 print("\n")
print("=" * 50)
print("STATISTICAL ANALYSIS FOR:", column)
print("=" * 50)
print("Mean:",dataFrame[column].mean())
print("Median:",dataFrame[column].median())
print("Mode:")
print(dataFrame[column].mode().head())
print("Standard Deviation:",dataFrame[column].std())
print("Variance:",dataFrame[column].var())
print("Minimum:",dataFrame[column].min())
print("Maximum:",dataFrame[column].max())
print("25th Percentile:",dataFrame[column].quantile(0.25))
print("50th Percentile:",dataFrame[column].quantile(0.50))
print("75th Percentile:",dataFrame[column].quantile(0.75))
print("Skewness:",dataFrame[column].skew())
print("Kurtosis:",dataFrame[column].kurt())
```
